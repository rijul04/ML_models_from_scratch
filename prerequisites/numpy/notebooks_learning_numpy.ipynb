**Adding, removing, and sorting elements**

This section covers np.sort(), np.concatenate()

```python
# Sorting with arrays

# Sorting with sort() - Returns a sorted copy of the passed in array
arr = np.array([2,6,1,4,8,3,2])
print(arr)
print(np.sort(arr))

# argsort - which is an indirect sort along a specified axis,

# lexsort - which is an indirect stable sort on multiple keys,

# searchsorted - which will find elements in a sorted array, and

# partition - which is a partial sort.
```

**How to convert a 1D array into a 2D array (how to add a new axis to an array)**

This section covers np.newaxis, np.expand_dims

```python
# np.newaxis to increase dimension by one

a = np.array([1,2,3,4])
print(a, a.shape)

print(a[np.newaxis, :], a[np.newaxis, :].shape)

print(a[:, np.newaxis], a[:, np.newaxis].shape)
```

```python
# np.expand_dims to expand array by inserting a new axis at a speicfied position

a = np.array([1,2,3,4])
print(a, a.shape)

print(np.expand_dims(a, axis=0), np.expand_dims(a, axis=0).shape)
print(np.expand_dims(a, axis=1), np.expand_dims(a, axis=1).shape)
```

**Indexing and slicing**

```python
# Slicing

arr = np.array([1,2,3,4])

print(arr[:2])
print(arr[2:])
print(arr[1:3])
print(arr[:-1])
```

```python
# Selecting values from array taht fulfill certain conditions

arr = np.arange(1, 13).reshape(3, 4)
print(arr)
print()

# Create a boolean mask to select certain values
print(arr > 6)
print()

print(arr[arr > 6])
print()

# And operations can be done with the & operator
print(arr[(arr>2) & (arr<10)])
print()

# Or operations can be done with the | operator
print(arr[(arr<5) | (arr==10)])
```

**How to create an array from existing data**

This section covers slicing and indexing, np.vstack(), np.hstack(), np.hsplit(), .view(), copy()

```python
# Stacking arrays

a = np.array([[1,1], [2,2]])
b = np.array([[3,3], [4,4]])


# Vertical stacking
print(np.vstack((a,b)))
print()

# Hozizontal stacking
print(np.hstack((a,b)))
```

```python
# Creating a View and Copy of an array

a = np.array([1,2,3,4])

# Copy
copy_a = a.copy()
copy_a[0] = 10
print(copy_a, a)
print()

# View (references the original array)
view_a = a.view()
view_a[0] = 10
print(view_a, a)
```

**How to save and load NumPy objects**

This section covers np.save, np.savez, np.savetxt, np.load, np.loadtxt

```python
# Storing a single ndarray object with .save() that stores data, shape, dtype of array

a = np.arange(5)

# Saves it as a .npy file
np.save("single_save", a)

# Load the .npy file
b = np.load("single_save.npy")
print(b)
```

```python
# Storing multiple ndarray object with .savez() that stores data, shape, dtype of array

a = np.arange(5)
b = np.arange(5, 10)

# Saves it as a .npz file
np.savez("multi_save", a=a, b=b)

# Load the .npy file
b = np.load("multi_save.npz")
print(b["a"], b["b"])
```

```python
# Basic version is storing numpy arrays as a plain text file like a .csv or .txt file using
# np.savetxt

a = np.arange(6)

np.savetxt("txt_save.csv", a)

txt = np.loadtxt("txt_save.csv")
print(txt)
```
