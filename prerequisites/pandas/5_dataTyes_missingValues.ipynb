```python
import pandas as pd

reviews = pd.read_csv("../../datasets/wine-reviews/winemag-data-130k-v2.csv", index_col=0)
reviews.columns
```

**Data Types**

The data type for a column in a DataFrame or a Series is known as the dtype

```python
# Getting the data type of a column in a DF

reviews["price"].dtype
```

```python
# Getting the data type for every column in a DF

reviews.dtypes
```

```python
# Converting dtype into another

reviews["points"].astype("float64")
```

```python
# Index of DF or Series has itws own dtype too

reviews.index.dtype
```

```python
# Pandas also support more exotic data types such as: categorical data and timeseries data
```

**Missing Data**

Entries missing values are given the value NaN - Not a Number which are of float64 dtype

pd.isnull() and pd.notnull() for selection based on missing values

```python
reviews[pd.isnull(reviews.country)]
```

```python
# Replace missing values

# Basic fill
print(reviews.region_2.fillna("unknown"))
```

```python
# Backfill (fills in with next available value)

reviews["price"].bfill()
```

```python
# Replace values

reviews["country"].replace("US", "United States")
```
