**Combining**

For combining different DataFramers and/or Series there are 3 main ways:
- .concat()
- .join()
- .merge() - Most of what merge does .join() can do so wont focus on it here



```python
# .concat() - Given a list of elements this func will join those elements together along an axis
# This is usefull when data is in different datasets but the fields (columns) are the same

# Joining 2 different countries youtube DF
canadian_yt = pd.read_csv("../../datasets/youtube/CAvideos.csv")
gb_yt = pd.read_csv("../../datasets/youtube/GBvideos.csv")

print(canadian_yt.shape)
print(gb_yt.shape)

pd.concat([canadian_yt, gb_yt]).shape

```

```python
# .join() - Combines different DF objects which have an index in common

left = canadian_yt.set_index(['title', 'trending_date'])
right = gb_yt.set_index(['title', 'trending_date'])

left.join(right, lsuffix='_CAN', rsuffix='_UK')
```
