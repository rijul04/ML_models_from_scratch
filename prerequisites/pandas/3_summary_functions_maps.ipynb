```python
import pandas as pd

reviews = pd.read_csv("../../datasets/wine-reviews/winemag-data-130k-v2.csv", index_col=0)
reviews.columns
```

**Summary functions**

Pandas provides simple summary funcs which restructure the data in some usefull way

```python
# Example summary func: .describe

print(reviews["points"].describe())
print()

# .describe is type dependant so int above vs string below
print(reviews["taster_name"].describe())
```

```python
# There are many but a few other basic examples:

# .mean
print(reviews["points"].mean())
print()

# .unique
print(reviews["taster_name"].unique())
print()

# .value_counts
print(reviews["taster_name"].value_counts())
```

**Maps**

Map - A function that takes a set of values and "maps" them to anotehr set of values

2 main mapping methods:
- .map()
- .apply()

Both return new Series and DF not transforming the original data

```python
# .map()
# Returns a series

reviews.points.map(lambda p: p - p)
```

```python
# .apply()
# The same as .map but with a function and returns a DataFrame instead

def remean_points(row):
    row.points = row.points - row.points
    return row

reviews = reviews.apply(remean_points, axis='columns')

reviews["points"]
```

```python
# There are common mapping operations that are faster such as the standard python operators
#  +, -, >, <, ==

mean_p = 50
reviews["price"] - mean_p
```
