```python
import pandas as pd
```

**Creating, Reading and Writing**

Creating:

There are 2 core objects in pandas:
- DataFrame
- Series


DataFrame - A table containing an array of entries where each entry responds to a row/record and a column which is a Series

Series - A sequence of data values like a list



**DataFrame**

```python
# Creating a DataFrame

# Data Frames take in a dictionary where keys are the column names 
# and the values are a list of entries
pd.DataFrame({"Yes": [100, 200], "No": [400, 500]})
```

```python
# Custom row labels (indexes) when creating a DF as by default its labels them as increasing numbers

pd.DataFrame({"Yes": [100, 200], "No": [400, 500]}, index=["Product A", "Product B"])
```

**Series**

```python
# Creating a Series

pd.Series([1,2,3,4,5])
```

```python
# Custom row labels (indexes)
pd.Series([1,2,3], index=["milk", "bread", "chocolate"], name="Items")
```

**Writing**

```python
# Writings a DF to a csv file

df = pd.DataFrame({"Col 1": [1, 2], "Col 2": [2, 4]}, index=["Row 1", "Row 2"])
print(df)
print()

df.to_csv("save.csv")

df_read = pd.read_csv("save.csv", index_col=0)
print(df_read)
```
