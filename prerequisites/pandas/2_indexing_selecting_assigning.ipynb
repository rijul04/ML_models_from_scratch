```python
import pandas as pd
```

```python
reviews = pd.read_csv("../../datasets/wine-reviews/winemag-data-130k-v2.csv", index_col=0)
reviews.columns
```

```python
# Accessing Column Properties from a df

reviews.country

# Above and Below are the same

reviews["country"]
```

```python
# Access a specific Row value from a column

reviews["country"][0]
```

**Indexing**

```python
# Index based selection uses .iloc

# Returns the first row of the df
print(reviews.iloc[0])
print()

# returns the first column of the df
print(reviews.iloc[:, 0])
print()

# Selecting specific rows with specific columns based on index
print(reviews.iloc[1:3, :3])
```

```python
# Label based selection using .loc

# [row number, column]
print(reviews.loc[0, "country"])
print()

# Returning rows with specific columns
print(reviews.loc[:, ["country", "taster_name"]])

```

```python
# Manipulate the index with .set_index

# sets title as the new index
reviews.set_index("title")
```

**Condintional Selecting**

```python
# Can compare values which returns a series of 
# True or False depending on the value (This is considered a mask)

reviews.country == "Italy"
```

```python
# The mask can then be used against the df for selection

reviews[reviews.country == "Italy"]
```

```python
# Can use the & or the | operator to combine masks

x = reviews[(reviews.country == 'Italy') & (reviews.points >= 90)]

y = reviews[(reviews.country == 'Italy') | (reviews.points >= 90)]

print(x.shape)
print()
print(y.shape)
```

```python
# Built in conditional selectors

# .isin
reviews[reviews.country.isin(["Italy", "France"])]

# .isnull and .notnull
reviews[reviews.price.isnull()]
reviews[reviews.price.notnull()]
```

**Assigning**

```python
# Assign constant value
reviews["critic"] = "me"
reviews["critic"]
```

```python
# assigning iterable of values

reviews['index_backwards'] = range(len(reviews), 0, -1)
reviews['index_backwards']
```
