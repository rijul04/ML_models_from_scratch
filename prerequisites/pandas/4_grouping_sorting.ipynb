```python
import pandas as pd

reviews = pd.read_csv("../../datasets/wine-reviews/winemag-data-130k-v2.csv", index_col=0)
reviews.columns
```

**Groupwise analysis**

Using .groupby() - Its creates groups (DataFrame like objects) for every group (a slice of the original DF)

```python
# Using .groupby() as .value_counts()

# .groupby() groups all rows that have the same value in the "points column"
#  then the points column is selected for each group
# .count() then counts the number of non-null values in the "points" column inside each group
reviews.groupby("points").points.count()
```

```python
# Any of the sumary funcs are available to use on grouped data

# Using .min()
print(reviews.groupby("points")["price"].min())
print()

# Using apply to get access to each sliced and grouped DF
reviews.groupby("taster_name").apply(lambda df: df.loc[df["points"].idxmax(), "title"])
```

```python
# grouping by more than one column
# doing this creates a multi-index for each grouping

reviews.groupby(["country", "points"])["price"].max()
```

```python
# Using .agg() summary func

reviews.groupby("points")["price"].agg([len, min, max])
```

**Multi-indexes**

when using groupby it can sometimes lead to multi idnexes when grouping by more than one column

```python
# Example of multi-indexed df

df = reviews.groupby(["country", "points"])["price"].max()
print(df)
print()

print(type(df.index))

```

```python
# converting multi-index back to regular index

df = df.reset_index()
df
```

**Sorting**

Generally returned data is displayed based on index order not value orders such as a value for a column

sort_values() allow sorting to happen

```python
# using .sort_values()

print(df)
print()

print(df.sort_values(by="price"))
```

```python
# sort by index values is same as .sort_values()

df.sort_index()
```
